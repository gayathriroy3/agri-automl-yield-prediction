```python
from google.cloud import aiplatform

# Initialize Vertex AI
aiplatform.init(
    project="your-project-id",
    location="us-central1",
    staging_bucket="gs://your-bucket-name",
)
```

```python
# Create a dataset
dataset = aiplatform.TabularDataset.create(
    display_name="crop-yield-dataset",
    gcs_source="gs://your-bucket-name/datasets/crop_yield_data.csv",
)
```

```python
# Train AutoML model
job = aiplatform.AutoMLTabularTrainingJob(
    display_name="crop-yield-train-job",
    optimization_prediction_type="regression",
    optimization_objective="minimize-rmse",
)

model = job.run(
    dataset=dataset,
    target_column="yield",
    model_display_name="crop-yield-model",
    budget_milli_node_hours=1000,
)
```

```python
# Deploy
endpoint = model.deploy(machine_type="n1-standard-2")

# Predict
response = endpoint.predict([
    {
        "soil_type": "clay",
        "rainfall": 260,
        "temperature": 27,
        "ph": 6.2,
        "crop_type": "wheat"
    }
])
print("Predicted Yield:", response.predictions)
```
